```python
import numpy as np

numeric_df = df.select_dtypes(include=[np.number])

Q1 = numeric_df.quantile(0.25)
Q3 = numeric_df.quantile(0.75)
IQR = Q3 -Q1

is_outlier = (numeric_df < (Q1 - 3.0 * IQR)) | (numeric_df > (Q3 + 3.0 * IQR))

rows_with_outliers = df[is_outlier.any(axis=1)]

clean_df  = df[~is_outlier.any(axis=1)]
```

```python
# Auditing the dropped rows..
original_count = len(df)
clean_count = len(clean_df)

dropped_count= original_count - clean_count

dropped_pct  = (dropped_count/original_count) * 100

print(f"Dropped {dropped_count} rows ({dropped_pct:.2f}% of the dataset).")
```

```python
# Calculating skewness for all numerical columns
numeric_df = df.select_dtypes(include=['number'])
skewness = numeric_df.skew().sort_values(ascending=False)

# Converting to a clean summary DataFrame
skew_df = pd.DataFrame({
    'skewness': skewness,
    'interpretation': skewness.apply(
        lambda x: 'Highly Right-Skewed' if x > 1 else (
                  'Highly Left-Skewed' if x < -1 else (
                  'Moderately Skewed' if abs(x) >= 0.5 else 'Symmetric')
        )
    )
})

print(skew_df)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

df.plot(
    y=['Infant_deaths', 'Under_five_deaths','Adult_mortality'],
    x='Year', 
    kind='line',
    figsize=(10, 6),                     
    color=['#2a9d8f', '#e9c46a', '#e76f51'], 
    linewidth=2,                         
    marker='o',                        
    title='Deaths Per Year'        
)

plt.ylabel('Year') 
plt.xlabel('Deaths')
plt.grid(True, linestyle='--', alpha=0.6) 
plt.show()
```

```python
# Mean of the years spanning all countries 
df_grouped = df.groupby('Year')[['Infant_deaths', 'Under_five_deaths', 'Adult_mortality']].mean()

df_grouped.plot(
    kind='line',
    figsize=(10, 6),
    color=['#2a9d8f', '#e9c46a', '#e76f51'],
    linewidth=2,
    marker='o',
    title='Average Deaths Per Year Globally'
)

plt.xlabel('Year')
plt.ylabel('Average Deaths')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
# Adult Mortality against Alcohol Consumption
df_grouped = df.groupby('Year')[['Alcohol_consumption', 'Adult_mortality']].mean()

df_grouped.plot(
    kind='line',
    figsize=(10, 6),
    color=['#2a9d8f', '#e9c46a', '#e76f51'],
    linewidth=2,
    marker='o',
    title='Relationship Between Alcohol Consumption and Adult Mortality Per Year'
)

plt.xlabel('Year')
plt.ylabel('Adult_mortality')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
