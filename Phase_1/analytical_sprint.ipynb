```python
import numpy as np

numeric_df = df.select_dtypes(include=[np.number])

Q1 = numeric_df.quantile(0.25)
Q3 = numeric_df.quantile(0.75)
IQR = Q3 -Q1

is_outlier = (numeric_df < (Q1 - 3.0 * IQR)) | (numeric_df > (Q3 + 3.0 * IQR))

rows_with_outliers = df[is_outlier.any(axis=1)]

clean_df  = df[~is_outlier.any(axis=1)]
```

```python
# Auditing the dropped rows..
original_count = len(df)
clean_count = len(clean_df)

dropped_count= original_count - clean_count

dropped_pct  = (dropped_count/original_count) * 100

print(f"Dropped {dropped_count} rows ({dropped_pct:.2f}% of the dataset).")
```

```python
# Calculating skewness for all numerical columns
numeric_df = df.select_dtypes(include=['number'])
skewness = numeric_df.skew().sort_values(ascending=False)

# Converting to a clean summary DataFrame
skew_df = pd.DataFrame({
    'skewness': skewness,
    'interpretation': skewness.apply(
        lambda x: 'Highly Right-Skewed' if x > 1 else (
                  'Highly Left-Skewed' if x < -1 else (
                  'Moderately Skewed' if abs(x) >= 0.5 else 'Symmetric')
        )
    )
})

print(skew_df)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

df.plot(
    y=['Infant_deaths', 'Under_five_deaths','Adult_mortality'],
    x='Year', 
    kind='line',
    figsize=(10, 6),                     
    color=['#2a9d8f', '#e9c46a', '#e76f51'], 
    linewidth=2,                         
    marker='o',                        
    title='Deaths Per Year'        
)

plt.ylabel('Year') 
plt.xlabel('Deaths')
plt.grid(True, linestyle='--', alpha=0.6) 
plt.show()
```

```python
# Mean of the years spanning all countries 
df_grouped = df.groupby('Year')[['Infant_deaths', 'Under_five_deaths', 'Adult_mortality']].mean()

df_grouped.plot(
    kind='line',
    figsize=(10, 6),
    color=['#2a9d8f', '#e9c46a', '#e76f51'],
    linewidth=2,
    marker='o',
    title='Average Deaths Per Year Globally'
)

plt.xlabel('Year')
plt.ylabel('Average Deaths')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
# Adult Mortality against Alcohol Consumption
df_grouped = df.groupby('Year')[['Alcohol_consumption', 'Adult_mortality']].mean()

df_grouped.plot(
    kind='line',
    figsize=(10, 6),
    color=['#2a9d8f', '#e9c46a', '#e76f51'],
    linewidth=2,
    marker='o',
    title='Relationship Between Alcohol Consumption and Adult Mortality Per Year'
)

plt.xlabel('Year')
plt.ylabel('Adult_mortality')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.regplot(
    data=df, 
    x='Alcohol_consumption', 
    y='Adult_mortality',
    scatter_kws={'alpha':0.4, 'color':'#2a9d8f'},
    line_kws={'color':'#e76f51', 'linewidth':2}
)

plt.title('Correlation: Alcohol Consumption vs. Adult Mortality')
plt.xlabel('Alcohol Consumption (Liters per capita)')
plt.ylabel('Adult Mortality (per 1,000)')
plt.grid(True, linestyle='--', alpha=0.6)
```

```python
corr_matrix = df.corr(numeric_only=True)
plt.figure(figsize=(6,10))
sns.heatmap(corr_matrix[['Life_expectancy']].sort_values(by='Life_expectancy', ascending=False), annot=True, cmap='coolwarm', fmt ='.2f', vmax=1, vmin=-1)
plt.title('WHO Life Expectancy Dataset Correlation Matrix')
```

```python
import scipy.stats as stats

developed = df[df['Economy_status_Developed'] == 1]['Life_expectancy']
developing = df[df['Economy_status_Developing'] == 1]['Life_expectancy']

stat, p_levene = stats.levene(developed, developing)
equal_variances = p_levene > 0.05

t_stat, p_val = stats.ttest_ind(developed, developing, equal_var=equal_variances)

print(f"Developed Mean Life Expectancy: {developed.mean():.2f} years")
print(f"Developing Mean Life Expectancy: {developing.mean():.2f} years")
print(f"T-statistic: {t_stat:.4f}")
print(f"P-value: {p_val:.4e}")

alpha = 0.05
if p_val < alpha:
    print("Result: Statistically significant difference. Reject the null hypothesis.")
else:
    print("Result: No statistically significant difference. Fail to reject the null hypothesis.")

```
