```python
import numpy as np

X = np.array([[234, 453],
                [42,533]])
```

```python
def sigmoid(z):
    g = 1/(1+np.exp(-z))
    return g
```

```python
g = sigmoid
```

```python
def my_dense(a_in, W, b):
    units = W.shape[1] # number of neurones
    a_out = np.zeros(units) # prepopulating the resulting matrix which will be the number of units since the results are getting squished by the neurones
    for j in range(units): # we're now interating through each neurone, great
        w = W[:, j] # selecting the weights for the jth neurone
        z = np.dot(a_in, w) + b[j] # each value will pass through the corresponging weight and bias
        a_out[j] = g(z)  # the resulting value is then passed through the sigmoid function
    return a_out
```

```python
def my_sequential(x, W1, b1, W2, b2):
    a1 = my_dense(x, W1, b1)
    a2 = my_dense(a1, W2, b2)
    return a2
```

```python

# Using some weights from the course
W1_tmp = np.array( [[-8.93,  0.29, 12.9 ], [-0.1,  -7.32, 10.81]] )
b1_tmp = np.array( [-9.82, -9.28,  0.96] )
W2_tmp = np.array( [[-31.18], [-27.59], [-32.56]] )
b2_tmp = np.array( [15.41] )

```

```python
# Flawed function
def my_predict(X, W1, b1, W2, b2):
    m = X.shape[0] # no of training samples.. each line.
    p = np.zeros((m,1)) # ?
    for i in range(m):
        p[i,0] = my_sequential(X[i], W1, b1, W2, b2)
    return(p)
```

```python
# Let's normalize with tensorflow
import tensorflow as tf
norm_l = tf.keras.layers.Normalization(axis=1)
```

```python
def new_my_predict(X, W1, b1, W2, b2):
    m = X.shape[0]
    p = np.zeros((m, 1))
    for i in range(m):
        res = my_sequential(X[i], W1, b1, W2, b2)
         # If 'res' is a tensor, convert to numpy first
        if hasattr(res, "numpy"):
            res = res.numpy()
        # Force conversion to a clean scalar float to prevent the sequence error
        p[i, 0] = float(np.squeeze(res))

    return p
norm_l.adapt(X)
X = norm_l(X)  # normalization
predictions = new_my_predict(X, W1_tmp, b1_tmp, W2_tmp, b2_tmp)
```

```python
y_hat = np.zeros_like(predictions)
for i in range(len(predictions)):
    if predictions[i] >= 0.5:
        y_hat[i] = 1
    else:
        y_hat[i] = 0
print(f'decisions = \n{y_hat}')
```
